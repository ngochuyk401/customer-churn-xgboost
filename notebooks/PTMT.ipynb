```python
# ================================
# IMPORT LIBRARIES
# ================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (8,6)
```

```python
# ================================
# LOAD DATA
# ================================

df = pd.read_csv('Churn_Cleaned_PTMT.csv')

# xem 5 dòng đầu
df.head()
```

```python
# Thống kê mô tả biến định lượng
print(df.describe())

# Thống kê mô tả biến định tính
print(df.describe(include='object'))
```

```python
df.columns
```

```python
# ================================
# NUMERICAL COLUMNS
# ================================

numerical_cols = ['tenure','MonthlyCharges','TotalCharges']
```

```python
# ================================
# HISTOGRAM + KDE
# ================================

for col in numerical_cols:

    plt.figure()

    sns.histplot(
        df[col],
        kde=True,
        bins=30
    )

    plt.title(f'Histogram và KDE của {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    plt.show()
```

```python
# ================================
# BOX PLOT
# ================================

for col in numerical_cols:

    plt.figure()

    sns.boxplot(
        y=df[col]
    )

    plt.title(f'Boxplot của {col}')

    plt.show()
```

```python
# ================================
# VIOLIN PLOT
# ================================

for col in numerical_cols:

    plt.figure()

    sns.violinplot(
        y=df[col]
    )

    plt.title(f'Violin plot của {col}')

    plt.show()
```

```python
# ================================
# CATEGORICAL COLUMNS
# ================================

categorical_cols = [
'Contract',
'InternetService',
'PaymentMethod',
'SeniorCitizen',
'TechSupport',
'OnlineSecurity',
'Churn'
]

# chỉ giữ cột tồn tại trong dataframe
categorical_cols = [col for col in categorical_cols if col in df.columns]

print(categorical_cols)
```

```python
# ================================
# BAR CHART
# ================================

for col in categorical_cols:

    plt.figure()

    sns.countplot(
        data=df,
        x=col,
        palette=sns.color_palette("Set2", df[col].nunique())
    )

    plt.title(f'Bar chart của {col}')

    plt.xticks(rotation=30)

    plt.show()
```

```python
# ================================
# PIE CHART cho các biến định tính
# ================================

for col in categorical_cols:

    if col in df.columns:

        counts = df[col].value_counts()

        plt.figure()

        plt.pie(
            counts,
            labels=counts.index,
            autopct='%1.1f%%',
            startangle=90
        )

        plt.title(f'Pie chart của {col}')

        plt.show()
```

```python
# ================================
# BOXPLOT THEO CHURN
# ================================

for col in numerical_cols:

    plt.figure()

    sns.boxplot(
        data=df,
        x='Churn',
        y=col
    )

    plt.title(f'{col} theo Churn')

    plt.show()
```

```python
# ================================
# VIOLIN + STRIP
# ================================

for col in numerical_cols:

    plt.figure()

    sns.violinplot(
        data=df,
        x='Churn',
        y=col
    )

    sns.stripplot(
        data=df,
        x='Churn',
        y=col,
        color='black',
        alpha=0.3
    )

    plt.title(f'Violin + Strip của {col} theo Churn')

    plt.show()
```

```python
# ================================
# STACKED BAR
# ================================

stack_cols = [
'Contract',
'InternetService',
'PaymentMethod',
'SeniorCitizen',
'TechSupport',
'OnlineSecurity'
]

stack_cols = [col for col in stack_cols if col in df.columns]

for col in stack_cols:

    cross_tab = pd.crosstab(
        df[col],
        df['Churn']
    )

    cross_tab.plot(
        kind='bar',
        stacked=True
    )

    plt.title(f'{col} theo Churn')

    plt.xlabel(col)
    plt.ylabel('Count')

    plt.xticks(rotation=30)

    plt.show()
```

```python
# ================================
# SCATTER PLOTS
# ================================

plt.figure()
sns.scatterplot(
    data=df,
    x='tenure',
    y='MonthlyCharges',
    hue='Churn',
    alpha=0.6
)
plt.title('Tenure vs MonthlyCharges')
plt.show()


plt.figure()
sns.scatterplot(
    data=df,
    x='tenure',
    y='TotalCharges',
    hue='Churn',
    alpha=0.6
)
plt.title('Tenure vs TotalCharges')
plt.show()


plt.figure()
sns.scatterplot(
    data=df,
    x='MonthlyCharges',
    y='TotalCharges',
    hue='Churn',
    alpha=0.6
)
plt.title('MonthlyCharges vs TotalCharges')
plt.show()
```

```python
# ================================
# CORRELATION HEATMAP
# ================================

corr = df[numerical_cols].corr(method='pearson')

plt.figure(figsize=(6,5))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt=".2f"
)

plt.title('Ma trận tương quan Pearson')

plt.show()
```
